```python
import os
from pigeon import annotate
from IPython.display import display, Image
```

```python
folder_path = '/content/Boots and Sneakers Dataset'
image_names = os.listdir(folder_path)
image_paths = [os.path.join(folder_path, image_name) for image_name in image_names]
```

```python
image_paths
```

```python
len(image_paths)
```

```python
label_options = ['boot', 'sneakers']
```

```python
from ast import Lambda
annotations = annotate(image_paths,options=label_options,display_fn=lambda filename:display(Image(filename)))
```

```python
import pandas as pd
df_annotation = pd.DataFrame(annotations,columns=['filename','label'])
df_annotation
```

```python
df_annotation.to_csv('boot_and_sneakers_annotation.csv', index=False)
```
